```python
import pandas as pd

pd.__version__
```

```python
from huggingface_hub import hf_hub_download

path = hf_hub_download(
    repo_id="McAuley-Lab/Amazon-Reviews-2023",
    filename="raw/review_categories/Magazine_Subscriptions.jsonl",
    repo_type="dataset",
    local_dir="data",
)
path
```

```python
reviews = pd.read_json(path, lines=True)
reviews.shape
```

```python
reviews.head()
```

```python
reviews.info()
```

```python
reviews.isna().sum()
```

```python
reviews.drop(columns=["images"]).duplicated().sum()
```

```python
dup_mask = reviews.drop(columns=["images"]).duplicated(keep=False)
reviews[dup_mask].sort_values(["user_id", "timestamp"]).head(10)
```

```python
(reviews["text"].str.strip() == "").sum(), (reviews["title"].str.strip() == "").sum()
```

```python
dup = reviews.drop(columns=["images"]).duplicated()
clean = reviews[~dup]
len(reviews), len(clean)
```

```python
clean = clean.assign(
    n_images=clean["images"].str.len().astype("int16"),
    rating=clean["rating"].astype("int8"),
    helpful_vote=clean["helpful_vote"].astype("int32"),
    year=clean["timestamp"].dt.year.astype("int16"),
).drop(columns=["images"])
clean.dtypes
```

```python
clean.to_parquet("data/processed/reviews", partition_cols=["year"], index=False)
```

```python
from pathlib import Path

jsonl_mb = Path(path).stat().st_size / 1024**2
parquet_files = Path("data/processed/reviews").rglob("*.parquet")
parquet_mb = sum(f.stat().st_size for f in parquet_files) / 1024**2
round(jsonl_mb, 1), round(parquet_mb, 1)
```
